```python
# ==========================================
# CREATE SEASON FEATURE
# ==========================================

def get_season(month):

    if month in [12, 1, 2]:
        return "Winter"

    elif month in [3, 4, 5]:
        return "Summer"

    elif month in [6, 7, 8, 9]:
        return "Monsoon"

    else:
        return "Post-Monsoon"


df["Season"] = df["Month"].apply(get_season)

df[["Date", "Month", "Season"]].head(10)
```

```python
# ==========================================
# PREPARE REGRESSION DATA
# ==========================================

regression_df = df.copy()

# Remove rows where target AQI is missing
regression_df = regression_df.dropna(subset=["AQI"])

print("Regression dataset shape:", regression_df.shape)
```

```python
# ==========================================
# SELECT FEATURES
# ==========================================

features = [
    "City",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "Season",
    "PM2.5",
    "PM10",
    "NO",
    "NO2",
    "NOx",
    "NH3",
    "CO",
    "SO2",
    "O3",
    "Benzene",
    "Toluene",
    "Xylene"
]

X = regression_df[features]

y = regression_df["AQI"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# ==========================================
# TIME-BASED SPLIT
# ==========================================

regression_df = regression_df.sort_values("Date")

n = len(regression_df)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

train_df = regression_df.iloc[:train_end]
val_df = regression_df.iloc[train_end:val_end]
test_df = regression_df.iloc[val_end:]

print("Training:", train_df.shape)
print("Validation:", val_df.shape)
print("Testing:", test_df.shape)

print("\nDate ranges:")

print("Training:",
      train_df["Date"].min(),
      "to",
      train_df["Date"].max())

print("Validation:",
      val_df["Date"].min(),
      "to",
      val_df["Date"].max())

print("Testing:",
      test_df["Date"].min(),
      "to",
      test_df["Date"].max())
```

```python
X_train = train_df[features]
y_train = train_df["AQI"]

X_val = val_df[features]
y_val = val_df["AQI"]

X_test = test_df[features]
y_test = test_df["AQI"]

print(X_train.shape)
print(X_val.shape)
print(X_test.shape)
```

```python
# ==========================================
# PREPROCESSING
# ==========================================

numeric_features = [
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "PM2.5",
    "PM10",
    "NO",
    "NO2",
    "NOx",
    "NH3",
    "CO",
    "SO2",
    "O3",
    "Benzene",
    "Toluene",
    "Xylene"
]

categorical_features = [
    "City",
    "Season"
]

numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
# ==========================================
# LINEAR REGRESSION
# ==========================================

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

linear_model.fit(X_train, y_train)

val_pred = linear_model.predict(X_val)

mae = mean_absolute_error(y_val, val_pred)
rmse = np.sqrt(mean_squared_error(y_val, val_pred))
r2 = r2_score(y_val, val_pred)

print("Linear Regression")
print("-------------------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R2  :", r2)
```

```python
# ==========================================
# RANDOM FOREST REGRESSION
# ==========================================

rf_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

rf_model.fit(X_train, y_train)

val_pred_rf = rf_model.predict(X_val)

mae_rf = mean_absolute_error(y_val, val_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_val, val_pred_rf))
r2_rf = r2_score(y_val, val_pred_rf)

print("Random Forest Regression")
print("-------------------------")
print("MAE :", mae_rf)
print("RMSE:", rmse_rf)
print("R2  :", r2_rf)
```

```python
# ==========================================
# GRADIENT BOOSTING REGRESSION
# ==========================================

gb_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=200,
                learning_rate=0.05,
                max_depth=3,
                random_state=42
            )
        )
    ]
)

gb_model.fit(X_train, y_train)

val_pred_gb = gb_model.predict(X_val)

mae_gb = mean_absolute_error(y_val, val_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_val, val_pred_gb))
r2_gb = r2_score(y_val, val_pred_gb)

print("Gradient Boosting Regression")
print("-------------------------")
print("MAE :", mae_gb)
print("RMSE:", rmse_gb)
print("R2  :", r2_gb)
```

```python
# ==========================================
# REGRESSION MODEL COMPARISON
# ==========================================
#For regression:

#Lower MAE/RMSE = better

#Higher R² = better

regression_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "Gradient Boosting"
    ],
    "MAE": [
        mae,
        mae_rf,
        mae_gb
    ],
    "RMSE": [
        rmse,
        rmse_rf,
        rmse_gb
    ],
    "R2": [
        r2,
        r2_rf,
        r2_gb
    ]
})

regression_results

```

```python
# ==========================================
# ACTUAL VS PREDICTED AQI
# ==========================================

best_regression_pred = val_pred_rf

plt.figure(figsize=(8, 8))

plt.scatter(
    y_val,
    best_regression_pred,
    alpha=0.5
)

plt.xlabel("Actual AQI")
plt.ylabel("Predicted AQI")

plt.title("Actual vs Predicted AQI")

# Perfect prediction line
min_value = min(y_val.min(), best_regression_pred.min())
max_value = max(y_val.max(), best_regression_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.show()
```

```python
# ==========================================
# CLASSIFICATION DATASET
# ==========================================

classification_df = df.copy()

# Remove rows where AQI category is missing
classification_df = classification_df.dropna(
    subset=["AQI_Bucket"]
)

print(
    "Classification dataset:",
    classification_df.shape
)

print(
    classification_df["AQI_Bucket"].value_counts()
)
```

```python
classification_features = [
    "City",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "Season",
    "PM2.5",
    "PM10",
    "NO",
    "NO2",
    "NOx",
    "NH3",
    "CO",
    "SO2",
    "O3",
    "Benzene",
    "Toluene",
    "Xylene"
]

classification_df = classification_df.sort_values("Date")

X = classification_df[classification_features]

y = classification_df["AQI_Bucket"]
```

```python
# ==========================================
# CLASSIFICATION TIME SPLIT
# ==========================================

n = len(classification_df)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

train_cls = classification_df.iloc[:train_end]
val_cls = classification_df.iloc[train_end:val_end]
test_cls = classification_df.iloc[val_end:]

X_train_cls = train_cls[classification_features]
y_train_cls = train_cls["AQI_Bucket"]

X_val_cls = val_cls[classification_features]
y_val_cls = val_cls["AQI_Bucket"]

X_test_cls = test_cls[classification_features]
y_test_cls = test_cls["AQI_Bucket"]

print("Training:", X_train_cls.shape)
print("Validation:", X_val_cls.shape)
print("Testing:", X_test_cls.shape)
```

```python
# ==========================================
# LOGISTIC REGRESSION
# ==========================================

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=2000
            )
        )
    ]
)

logistic_model.fit(
    X_train_cls,
    y_train_cls
)

pred_logistic = logistic_model.predict(
    X_val_cls
)

accuracy = accuracy_score(
    y_val_cls,
    pred_logistic
)

precision = precision_score(
    y_val_cls,
    pred_logistic,
    average="weighted"
)

recall = recall_score(
    y_val_cls,
    pred_logistic,
    average="weighted"
)

f1 = f1_score(
    y_val_cls,
    pred_logistic,
    average="weighted"
)

print("Logistic Regression")
print("-------------------------")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

```python
# ==========================================
# DECISION TREE CLASSIFIER
# ==========================================

dt_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            DecisionTreeClassifier(
                max_depth=10,
                random_state=42
            )
        )
    ]
)

dt_model.fit(
    X_train_cls,
    y_train_cls
)

pred_dt = dt_model.predict(
    X_val_cls
)

accuracy_dt = accuracy_score(
    y_val_cls,
    pred_dt
)

f1_dt = f1_score(
    y_val_cls,
    pred_dt,
    average="weighted"
)

print("Decision Tree")
print("-------------------------")
print("Accuracy:", accuracy_dt)
print("F1 Score:", f1_dt)
```

```python
# ==========================================
# RANDOM FOREST CLASSIFIER
# ==========================================

rf_classifier = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestClassifier(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

rf_classifier.fit(
    X_train_cls,
    y_train_cls
)

pred_rf_cls = rf_classifier.predict(
    X_val_cls
)

accuracy_rf_cls = accuracy_score(
    y_val_cls,
    pred_rf_cls
)

precision_rf_cls = precision_score(
    y_val_cls,
    pred_rf_cls,
    average="weighted"
)

recall_rf_cls = recall_score(
    y_val_cls,
    pred_rf_cls,
    average="weighted"
)

f1_rf_cls = f1_score(
    y_val_cls,
    pred_rf_cls,
    average="weighted"
)

print("Random Forest Classifier")
print("-------------------------")
print("Accuracy :", accuracy_rf_cls)
print("Precision:", precision_rf_cls)
print("Recall   :", recall_rf_cls)
print("F1 Score :", f1_rf_cls)
```

```python
# ==========================================
# CLASSIFICATION MODEL COMPARISON
# ==========================================

classification_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        accuracy,
        accuracy_dt,
        accuracy_rf_cls
    ],
    "F1 Score": [
        f1,
        f1_dt,
        f1_rf_cls
    ]
})

classification_results
```

```python
depths = [5, 10, 15, 20, None]

depth_results = []

for depth in depths:

    model = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", RandomForestClassifier(
                n_estimators=150,
                max_depth=depth,
                random_state=42,
                n_jobs=-1
            ))
        ]
    )

    model.fit(X_train_cls, y_train_cls)

    prediction = model.predict(X_val_cls)

    acc = accuracy_score(y_val_cls, prediction)

    f1_value = f1_score(
        y_val_cls,
        prediction,
        average="weighted"
    )

    depth_results.append([
        depth,
        acc,
        f1_value
    ])

depth_results_df = pd.DataFrame(
    depth_results,
    columns=["max_depth", "Accuracy", "F1"]
)

depth_results_df
```

# **SVM**

```python
# ==========================================
# SUPPORT VECTOR MACHINE (SVM) CLASSIFIER
# ==========================================

from sklearn.svm import SVC

svm_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            SVC(
                kernel="rbf",
                C=1.0,
                gamma="scale"
            )
        )
    ]
)

# Train the SVM model
svm_model.fit(X_train_cls, y_train_cls)

# Validation predictions
pred_svm = svm_model.predict(X_val_cls)

# Evaluation
accuracy_svm = accuracy_score(y_val_cls, pred_svm)
precision_svm = precision_score(
    y_val_cls,
    pred_svm,
    average="weighted"
)
recall_svm = recall_score(
    y_val_cls,
    pred_svm,
    average="weighted"
)
f1_svm = f1_score(
    y_val_cls,
    pred_svm,
    average="weighted"
)

print("Support Vector Machine (SVM)")
print("-------------------------")
print("Accuracy :", accuracy_svm)
print("Precision:", precision_svm)
print("Recall   :", recall_svm)
print("F1 Score :", f1_svm)
```

# **KNN**

```python
# ==========================================
# K-NEAREST NEIGHBORS (KNN)
# ==========================================

from sklearn.neighbors import KNeighborsClassifier

# Numeric preprocessing with scaling
numeric_transformer_scaled = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical preprocessing
categorical_transformer_scaled = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

knn_preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer_scaled, numeric_features),
        ("cat", categorical_transformer_scaled, categorical_features)
    ]
)

knn_model = Pipeline(
    steps=[
        ("preprocessor", knn_preprocessor),
        (
            "model",
            KNeighborsClassifier(
                n_neighbors=5,
                weights="distance"
            )
        )
    ]
)

# Train
knn_model.fit(X_train_cls, y_train_cls)

# Validation prediction
pred_knn = knn_model.predict(X_val_cls)

# Evaluation
accuracy_knn = accuracy_score(
    y_val_cls,
    pred_knn
)

precision_knn = precision_score(
    y_val_cls,
    pred_knn,
    average="weighted"
)

recall_knn = recall_score(
    y_val_cls,
    pred_knn,
    average="weighted"
)

f1_knn = f1_score(
    y_val_cls,
    pred_knn,
    average="weighted"
)

print("K-Nearest Neighbors")
print("-------------------------")
print("Accuracy :", accuracy_knn)
print("Precision:", precision_knn)
print("Recall   :", recall_knn)
print("F1 Score :", f1_knn)
```

# **Naive Bayes**

```python
# ==========================================
# GAUSSIAN NAIVE BAYES
# ==========================================

from sklearn.naive_bayes import GaussianNB

nb_numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

nb_categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

nb_preprocessor = ColumnTransformer(
    transformers=[
        ("num", nb_numeric_transformer, numeric_features),
        ("cat", nb_categorical_transformer, categorical_features)
    ]
)

nb_model = Pipeline(
    steps=[
        ("preprocessor", nb_preprocessor),
        ("model", GaussianNB())
    ]
)

# Train
nb_model.fit(X_train_cls, y_train_cls)

# Validation prediction
pred_nb = nb_model.predict(X_val_cls)

# Evaluation
accuracy_nb = accuracy_score(
    y_val_cls,
    pred_nb
)

precision_nb = precision_score(
    y_val_cls,
    pred_nb,
    average="weighted"
)

recall_nb = recall_score(
    y_val_cls,
    pred_nb,
    average="weighted"
)

f1_nb = f1_score(
    y_val_cls,
    pred_nb,
    average="weighted"
)

print("Gaussian Naive Bayes")
print("-------------------------")
print("Accuracy :", accuracy_nb)
print("Precision:", precision_nb)
print("Recall   :", recall_nb)
print("F1 Score :", f1_nb)
```
